```python
#Membuat array 2D
matriks=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print(matriks)
```

```python
#Mengakses array 2D =>array[baris][kolom]

print(matriks[1][1])
print(matriks[2][0])
```

```python
#Menampilkan seluruh data array 2D secara manual
print(matriks[0][0])
print(matriks[0][1])
print(matriks[0][2])

print(matriks[1][0])
print(matriks[1][1])
print(matriks[1][2])

print(matriks[2][0])
print(matriks[2][1])
print(matriks[2][2])
```

```python
#Menampilkan seluruh data array 2D menggunakan looping range
for baris in range(3):
  for kolom in range(3):
    print(f"{matriks[baris][kolom]}", end=" ")
  print()
```

```python
#Menampilkan seluruh data array 2D menggunakan looping list

for baris in matriks:
  for kolom in baris:
    print(kolom, end=" ")
  print()
```

```python
def show_matriks(matrik):
  for baris in matrik:
    for kolom in baris:
      print(kolom, end=" ")
    print()
```

```python
show_matriks(matriks)
```

```python
#Merubah nilai array 2D ==>array[baris][kolom]=new_value

#Mengganti angka 5 menjadi 50
matriks[1][1]=50

#Mengganti angka 6 menjadi 30
matriks[1][2]=30

#Mengganti angka 3 menjadi 10
matriks[0][2]=10

#Mengganti angka 7 menjadi 20
matriks[2][0]=20

show_matriks(matriks)
```

```python
#Input Array 2D
#Data Tabel Nilai Mahasiswa
#Nama         Nilai
#Ade Levi       80
#Amelia         60
#Dini Delsa     90
#Luthfi         100

nilai_mhs=[
    ["Ade Levi",80],
    ["Amelia",60],
    ["Dini Delsa",90],
    ["Luthfi",100]
]

show_matriks(nilai_mhs)
```

```python
tabel_nilai_mhs=[]
jum_baris=4
jum_kolom=2

for baris in range(jum_baris):
  print(f"Masukkan nilai mahasiswa ke-{baris+1}")
  print("Data ke-1=Nama, Data Ke-2=Nilai)")
  data_baris=[]
  for kolom in range(jum_kolom):
    data=input(f"Data ke-{kolom+1} ?")
    data_baris.append(data)
  print()
  tabel_nilai_mhs.append(data_baris)


show_matriks(tabel_nilai_mhs)
```
