```python
matrik_A=[
    [3,1],
    [5,2]]

a=matrik_A[0][0]
b=matrik_A[0][1]
c=matrik_A[1][0]
d=matrik_A[1][1]

adj_a=[
    [d,-b],
    [-c,a]
]

x=1/((a*d)-(b*c))

#Invers_A=x*adj_a
```

```python
Invers_A = [[0, 0], [0, 0]]

for i in range(2):
  for j in range(2):
    Invers_A[i][j]=x*adj_a[i][j]


for i in Invers_A:
  for j in i :
    print(j, end=" ")
  print()
```

```python
def inverse_matrix(matrix):
    n = len(matrix)

    # Membuat matriks identitas
    identity = [[float(i == j) for j in range(n)] for i in range(n)]

    # Gabungkan matrix dan identity
    augmented = [row[:] + identity_row[:] for row, identity_row in zip(matrix, identity)]

    # Proses Gauss-Jordan
    for i in range(n):

        # Cari pivot
        pivot = augmented[i][i]

        if pivot == 0:
            raise ValueError("Matriks tidak memiliki invers")

        # Membagi baris agar pivot = 1
        for j in range(2 * n):
            augmented[i][j] /= pivot

        # Eliminasi baris lain
        for k in range(n):
            if k != i:
                factor = augmented[k][i]

                for j in range(2 * n):
                    augmented[k][j] -= factor * augmented[i][j]

    # Ambil bagian kanan sebagai invers
    inverse = [row[n:] for row in augmented]

    return inverse
```

```python
A = [
    [4, 7],
    [2, 6]
]

inv = inverse_matrix(A)

for row in inv:
    print(row)
```
