```python
#Membuat sebuat prosedur/fungsi tanpa parameter di python
#def nama_prosedur():
#  <Isi Prosedur >

#def nama_fungsi():
#  <Isi Fungsi >
#  return value

def say_hello():
  print("Hallo selamat belajar python..")

#Untuk memanggil prosedur/fungsi
# nama_fungsi/prosedur(optional params)

say_hello()

def nilai_pi():
  return 3.14

def gravitasi():
  return 9.8

jari_lingkaran=20
luas_lingkaran=nilai_pi()*jari_lingkaran**2

masa_benda=112
gaya_berat=masa_benda*gravitasi()

print(f"Luas lingkaran dengan r= {jari_lingkaran} adalah {luas_lingkaran:,.2f}")
print(f"Gaya berat benda dengan masa benda= {masa_benda} adalah {gaya_berat:,.3f}")
```

```python
#Prosedur dengan parameter

def say_hello_to(nama):
  print(f"Hallo {nama}, selamat belajar python")

def luas_lingkaran(r):
  return nilai_pi()*r**2


say_hello_to("Ade Levi Kurniawan")
say_hello_to("Amelia Puspitasari")

r1=12
r2=15
print(f"Luas lingkaran dengan r={r1} adalah {luas_lingkaran(r1):,.2f}")
print(f"Luas lingkaran dengan r={r2} adalah {luas_lingkaran(r2):,.2f}")
```

# Rumus Luas Permukaan Tabung
## L = 2 π r (r + t)


```python
def luas_permukaan_tabung(r,t):
  return 2*nilai_pi()*r*(r+t)

def d2r(d):
  return 0.5*d
```

1) Seorang perajin kayu ingin membuat meja dari batang pohon yang berbentuk tabung dengan panjang diameter 28 cm dan tinggi 50 cm. Berapakah luas permukaan dari batang kayu tersebut?

```python
diameter=28
tinggi=50

jari2=d2r(diameter)

luas_pt=luas_permukaan_tabung(jari2,tinggi)

print(f"Luas permukaan tabung dengan diameter ={diameter}cm dan tinggi= {tinggi}cm adalah {luas_pt:,.2f} persegi")
```

## Fungsi dengan return value lebih dari satu

```python
def luas_persegi_panjang(p,l):
  return p*l

def kel_persegi_panjang(p,l):
  return 2*(p+l)

def persegi_panjang(p,l):
  return p*l,2*(p+l)
```

```python
panjang=10
lebar=5
print(f"Luas PP : {luas_persegi_panjang(panjang,lebar)}")

print(f"Keliling PP : {kel_persegi_panjang(panjang,lebar)}")
```

```python
panjang,lebar=10,5

luas,keliling=persegi_panjang(panjang,lebar)
print(f"Luas PP : {luas}")
print(f"Keliling PP : {keliling}")
```
